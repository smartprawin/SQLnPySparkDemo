# PySpark 2: Transformations vs Actions (Lazy Evaluation)

```python
from pyspark.sql import SparkSession, functions as F
spark = SparkSession.builder.master("local[*]").appName("t-vs-a").getOrCreate()

users = spark.createDataFrame(
    [(1, "alice", 30, "US"),
     (2, "bob", 17, "CA"),
     (3, "carol", 25, "US")],
    ["id", "name", "age", "country"])

df2 = users.filter(F.col("age") > 18)   # transformation
df3 = df2.select("name")                # transformation

df3.explain()   # shows the optimized (lazy) plan; no data scanned yet
df3.show()      # action -> triggers execution
```

```python
spark.stop()
```
